```python
import astropy.constants as ct
from math import sqrt
```

```python
mp = float(ct.m_p.cgs.value)
me = float(ct.m_e.cgs.value)
kb = float(ct.k_B.cgs.value)
mu = 0.5*(mp + me)
Gamma = 5/3
```

```python
n1 = 2*3.7e-3   # electrons + protons
T1 = 9.3e7
n2 = 2*9.3e-3
T2 = 2.2e8
```

```python
p1 = n1*kb*T1
p2 = n2*kb*T2
print(p1, p2)
```

```python
M1 = sqrt(1/(2*Gamma)*(Gamma*(1 + p2/p1) + (p2/p1 - 1)))
print(M1)
```

```python
c1 = sqrt(Gamma*kb*T1/mu)
print(c1/1e5)
```

```python
u1 = M1*c1
print(u1/1e5)
```

```python
u2 = (n1/n2)*u1
print(u2/1e5)
```

```python
rho1 = n1*mu
rho2 = n2*mu
V1 = 1/rho1
V2 = 1/rho2
```

```python
j = sqrt((p2 - p1)/(V1 - V2))
print(j)
```
